```python
import nltk
import string
import pandas as pd

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

print("All libraries imported successfully!")
```

```python
# Creating a sample text dataset

data = {
    'Text': [
        "I am learning Natural Language Processing!",
        "Machine Learning is very interesting.",
        "I love Python and Artificial Intelligence.",
        "NLP helps computers understand human language.",
        "Data Science is an exciting field."
    ]
}

df = pd.DataFrame(data)

# Display the original dataset
print("Original Text Dataset:")
display(df)
```

```python
# Step 3: Tokenization

df['Tokens'] = df['Text'].apply(word_tokenize)

# Display Before and After
print("Original Text vs Tokenized Text")

display(df[['Text', 'Tokens']])
```

```python
# Step 4: Stopwords Removal

# Get English stopwords
stop_words = set(stopwords.words('english'))

# Remove stopwords from tokenized text
df['Filtered_Tokens'] = df['Tokens'].apply(
    lambda tokens: [word for word in tokens if word.lower() not in stop_words]
)

# Display Before and After
print("Before and After Stopwords Removal:")

display(df[['Tokens', 'Filtered_Tokens']])
```

```python
# Step 5: Punctuation Removal

# Remove punctuation marks from filtered tokens
df['Cleaned_Tokens'] = df['Filtered_Tokens'].apply(
    lambda tokens: [
        word for word in tokens
        if word not in string.punctuation
    ]
)

# Convert token lists back into sentences
df['Cleaned_Text'] = df['Cleaned_Tokens'].apply(
    lambda tokens: ' '.join(tokens)
)

# Display Before and After
print("Before and After Punctuation Removal:")

display(df[['Filtered_Tokens', 'Cleaned_Tokens', 'Cleaned_Text']])
```

```python
# Step 6: Bag of Words (BoW)

# Initialize CountVectorizer
vectorizer = CountVectorizer()

# Convert cleaned text into numerical vectors
bow_matrix = vectorizer.fit_transform(df['Cleaned_Text'])

# Get all unique words (Vocabulary)
feature_names = vectorizer.get_feature_names_out()

# Convert BoW matrix into a DataFrame
bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=feature_names
)

# Display Vocabulary
print("Vocabulary:")
print(feature_names)

# Display Bag of Words Representation
print("\nBag of Words Matrix:")
display(bow_df)
```

```python
# Step 7: Export Preprocessed Dataset

# Create final dataset
final_df = pd.DataFrame({
    'Original_Text': df['Text'],
    'Tokenized_Text': df['Tokens'].apply(lambda x: ' '.join(x)),
    'Stopwords_Removed': df['Filtered_Tokens'].apply(lambda x: ' '.join(x)),
    'Cleaned_Text': df['Cleaned_Text']
})

# Save preprocessed dataset as CSV
final_df.to_csv('preprocessed_dataset.csv', index=False)

# Display final dataset
print("Final Preprocessed Dataset:")
display(final_df)

print("\nDataset exported successfully!")
```

```python
# Create the complete preprocessed dataset

complete_dataset = pd.concat(
    [final_df, bow_df],
    axis=1
)

# Export dataset as CSV
complete_dataset.to_csv(
    'preprocessed_dataset.csv',
    index=False
)

print("Dataset exported successfully!")

display(complete_dataset.head())
```
